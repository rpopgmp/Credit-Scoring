Распределения плотности для процентной ставки и возраста в целом похожи между положительным и отрицательным классами: у обоих классов плотности перекрываются.

Для величины кредита видно, что у положительного класса более высокий пик на меньших суммах, а распределение отрицательного класса обладает более тяжёлым правым хвостом. Это интерпретируется так: небольшие кредиты чаще возвращаются, а с ростом суммы кредита вероятность дефолта увеличивается.

Аналогично для срока кредита: оба класса имеют моду при относительно коротких сроках, но у отрицательного класса распределение шире и на больших сроках его плотность больше — то есть длительные кредиты ассоциируются с более высоким риском невозврата.

#Классификация

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
data['Telephone'] = data['Telephone'].map({'yes': 1, 'none': 0})
data['Foreign worker'] = data['Foreign worker'].map({'yes': 1, 'no': 0})
data['Status of existing checking account'] = data['Status of existing checking account'].map({'good': 1, 'bad': 0})
```

```python
categorical = [
    'Credit history',
    'Purpose of the credit',
    'Status of savings account/bonds',
    'Present employment(years)',
    'personal_status',
    'Other debtors / guarantors',
    'Present residence since X years',
    'Property',
    'Other installment plans (banks/stores)',
    'Housing',
    'Job'
    ]

numeric = [
    'Duration in months',
    'Credit amount',
    'Installment rate in percentage of disposable income',
    'Age in years',
    'Number of existing credits at this bank'
]
```

Отбор признаков

```python
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    precision_score, recall_score, roc_auc_score,
    roc_curve, f1_score, accuracy_score
)
from mlxtend.feature_selection import SequentialFeatureSelector as SFS
from sklearn.model_selection import train_test_split


def preprocess(data=data, target='Status of existing checking account', categorical=categorical, numeric=numeric):
    X = data.drop(target, axis=1)
    y = data[target]

    numeric_transformer = StandardScaler()
    categorical_transformer = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

    preprocessor = ColumnTransformer(
        transformers=[
            ('num', numeric_transformer, numeric),
            ('cat', categorical_transformer, categorical)
        ]
    )

    X_preprocessed = preprocessor.fit_transform(X)

    return X_preprocessed, y, preprocessor


X, y, preprocessor = preprocess()

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2,shuffle=True, random_state=42)
```

```python
def get_feature_names(preprocessor):
    feature_names = []

    for name, transformer, cols in preprocessor.transformers_:
        if name == 'num':  # числовые признаки
            feature_names.extend(cols)
        elif name == 'cat':  # категориальные
            ohe = transformer
            ohe_features = ohe.get_feature_names_out(cols)
            feature_names.extend(ohe_features)
    return feature_names

feature_names = get_feature_names(preprocessor)
```

```python
rf = RandomForestClassifier(n_estimators=100, random_state=42)
rf.fit(X_train, y_train)
```

```python
importances = rf.feature_importances_
importance_df = pd.DataFrame({
    'feature': feature_names,
    'importance': importances
}).sort_values(by='importance', ascending=False)

top_features = importance_df['feature'].iloc[:20].tolist()
```

```python
importance_df
```

```python
from mlxtend.feature_selection import SequentialFeatureSelector as SFS

X_pre_df = pd.DataFrame(X_train, columns=feature_names)

X_top = X_pre_df[top_features]

sfs = SFS(rf,
          k_features='best',
          forward=True,
          floating=False,
          scoring='roc_auc',
          cv=5,
          n_jobs=-1)

sfs = sfs.fit(X_top, y_train)

selected_features = list(sfs.k_feature_names_)
```

```python
selected_features
```

Классификация

```python
from catboost import CatBoostClassifier

X_selected = X_pre_df[selected_features]
y_target = y_train

X_pre_df_test = pd.DataFrame(X_test, columns=feature_names)
X_test_sel = X_pre_df_test[selected_features]


cat_model = CatBoostClassifier(
    iterations=1500,
    learning_rate=0.05,
    depth=6,
    eval_metric='AUC',
    random_seed=42,
    verbose=100,
    class_weights=[2, 1]
)

cat_model.fit(X_selected, y_train, eval_set=(X_test_sel, y_test), use_best_model=True)



y_pred = cat_model.predict(X_test_sel)

y_proba = cat_model.predict_proba(X_test_sel)[:, 1]

accuracy = accuracy_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)
precision = precision_score(y_test, y_pred)
recall = recall_score(y_test, y_pred)
roc_auc = roc_auc_score(y_test, y_proba)


print(f"Accuracy: {accuracy:.4f}")
print(f"F1-score: {f1:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"ROC-AUC: {roc_auc:.4f}")

```

```python
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
import matplotlib.pyplot as plt

plt.figure(figsize=(6,6))
plt.plot(fpr, tpr, label=f'CatBoost (AUC = {roc_auc:.4f})')
plt.plot([0,1], [0,1], 'k--', label='Random')
plt.xlabel('FPR')
plt.ylabel('TPR')
plt.title('ROC Кривая')
plt.legend()
plt.show()
```
